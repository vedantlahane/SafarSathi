```python
import os
import pandas as pd

def inspect_raw_data(base_path):
    print(f"Inspecting data in: {base_path}\n" + "="*50)

    for root, dirs, files in os.walk(base_path):
        if '.ipynb_checkpoints' in root:
            continue

        csv_files = [f for f in files if f.endswith('.csv')]

        if csv_files:
            print(f"\nDirectory: {root}")
            print("-" * 30)

            for file in csv_files:
                file_path = os.path.join(root, file)
                size_mb = os.path.getsize(file_path) / (1024 * 1024)

                try:
                    # Load first 5 rows
                    df_preview = pd.read_csv(file_path, nrows=5)

                    print(f"\n📄 File: {file} ({size_mb:.2f} MB)")
                    print(f"Features: {df_preview.columns.tolist()}")
                    print("Preview:")
                    display(df_preview)
                    print("." * 40)
                except Exception as e:
                    print(f"  Error reading {file}: {e}")

raw_data_path = '/content/drive/MyDrive/yatrax-ml/data/raw'

if os.path.exists(raw_data_path):
    inspect_raw_data(raw_data_path)
else:
    print(f"Path not found: {raw_data_path}. Check your Drive structure.")
```
